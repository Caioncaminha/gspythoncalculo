## Modelagem da Curva de Aprendizado com Cálculo

Nesta fase, modelamos o processo de aprendizado contínuo e aquisição de habilidades ao longo do tempo. A função escolhida representa um aprendizado inicial rápido, seguido por uma estabilização à medida que o conhecimento se aproxima de uma capacidade máxima.

Usaremos a função de crescimento exponencial até um limite como um modelo adequado para curvas de aprendizado:
$$ K(t) = L(1 - e^{-kt}) $$

Onde:
- $K(t)$: Nível de conhecimento adquirido no tempo $t$
- $t$: Tempo dedicado ao estudo (ex: em meses ou semanas)
- $L$: O nível máximo de conhecimento possível (o limite superior)
- $k$: A taxa de aprendizado, indicando a rapidez com que o conhecimento é adquirido

Analisaremos esta função usando conceitos-chave de cálculo:
- **Limites:** Para entender o conhecimento potencial máximo ($L$).
- **Derivadas:** Para medir a velocidade do aprendizado ($K'(t)$).
- **Integrais:** Para estimar o conhecimento total acumulado ao longo de um período (conceitual).

Vamos escolher $L=100$ (representando 100% da capacidade de conhecimento) e definir uma taxa de aprendizado base $k=0.2$ para a exploração inicial. Posteriormente, $k$ será vinculado ao nosso 'Índice de Prontidão para o Futuro (IPF)'.

```python
import numpy as np
import matplotlib.pyplot as plt

def funcao_conhecimento(t, L, k):
    return L * (1 - np.exp(-k * t))

# --- Análise com Cálculo ---

# 1. Limite quando t tende ao infinito
L_val = 100
print(f"\n1. Análise de Limite para K(t):")
print(f"   Quando t tende ao infinito, K(t) se aproxima de L. Portanto, lim (t->inf) K(t) = {L_val}")
print(f"   Isso representa o nível máximo de conhecimento ou habilidade atingível.")

# 2. Derivada K'(t) para encontrar a velocidade do aprendizado
def funcao_velocidade_aprendizado(t, L, k):
    return L * k * np.exp(-k * t)

print(f"\n2. Análise da Derivada para K'(t) (Velocidade de Aprendizado):")
print(f"   K'(t) = L * k * e^(-kt)")
print(f"   Esta função representa a rapidez com que novo conhecimento é adquirido. Inicialmente, quando t é pequeno, e^(-kt) está próximo de 1, então K'(t) é alto (aprendizado rápido). Conforme t aumenta, e^(-kt) se aproxima de 0, então K'(t) se aproxima de 0 (o aprendizado desacelera).")

# 3. Integração Conceitual de K(t)
print(f"\n3. Análise da Integral (Conceitual):")
print(f"   A integral de K(t) em relação a t representa o conhecimento total acumulado ao longo de um determinado intervalo de tempo.")
print(f"   Matematicamente, a integral de K(t) de 0 a T é: L * [t + (1/k) * e^(-kt)] de 0 a T.")
print(f"   = L * [T + (1/k) * e^(-kT) - (1/k)]")
print(f"   Isso fornece o 'volume' total de conhecimento acumulado ao longo do tempo T.")
```

```python
# Integração e Visualização Dinâmica
import ipywidgets as widgets
from ipywidgets import interact

def plotar_curva_aprendizado(nome_pais):
    """
    Plota a curva de aprendizado e a velocidade do aprendizado para um país selecionado,
    com a taxa de aprendizado 'k' ajustada dinamicamente pelo IPF do país.
    """
    if 'df_final' not in globals() or df_final is None:
        print("O DataFrame 'df_final' não está disponível. Por favor, execute as células anteriores primeiro.")
        return

    dados_pais = df_final[df_final['Pais'] == nome_pais].iloc[0]
    valor_ipf = dados_pais['IPF']
    
    k_min = 0.1
    k_max = 0.5
    k_dinamico = k_min + (valor_ipf * (k_max - k_min))
    
    L = 100
    t = np.linspace(0, 50, 500)
    
    conhecimento = funcao_conhecimento(t, L, k_dinamico)
    velocidade_aprendizado = funcao_velocidade_aprendizado(t, L, k_dinamico)
    
    plt.style.use('seaborn-v0_8-whitegrid')
    fig, ax1 = plt.subplots(figsize=(12, 7))

    cor = 'tab:blue'
    ax1.set_xlabel('Tempo (ex: semanas de estudo)')
    ax1.set_ylabel('Conhecimento Adquirido (%)', color=cor)
    ax1.plot(t, conhecimento, color=cor, label='Conhecimento K(t)')
    ax1.tick_params(axis='y', labelcolor=cor)
    ax1.set_ylim(0, L + 5)

    ax2 = ax1.twinx()
    cor = 'tab:red'
    ax2.set_ylabel('Velocidade de Aprendizado', color=cor)
    ax2.plot(t, velocidade_aprendizado, color=cor, linestyle='--', label="Velocidade K'(t)")
    ax2.tick_params(axis='y', labelcolor=cor)

    plt.title(f'Curva de Aprendizado para {nome_pais}\n(IPF: {valor_ipf:.2f} -> Taxa de Aprendizado k: {k_dinamico:.2f})', fontsize=16)
    fig.tight_layout()
    
    linhas, legendas = ax1.get_legend_handles_labels()
    linhas2, legendas2 = ax2.get_legend_handles_labels()
    ax2.legend(linhas + linhas2, legendas + legendas2, loc='center right')
    
    plt.show()

if 'df_final' in globals() and df_final is not None:
    lista_paises = sorted(df_final['Pais'].unique())
    widget_paises = widgets.Dropdown(options=lista_paises, value='Australia', description='País:')
    interact(plotar_curva_aprendizado, nome_pais=widget_paises)
else:
    print("O DataFrame 'df_final' não está disponível para criar o widget interativo.")
```
